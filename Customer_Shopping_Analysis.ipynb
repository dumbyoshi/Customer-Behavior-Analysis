```python
dataset['Review Rating'] = dataset.groupby('Category')['Review Rating'] . transform (lambda x : x.fillna(x.median()))
```

```python
dataset.isnull().sum()
```

```python
dataset.columns = dataset.columns.str.lower()
dataset.columns = dataset.columns.str.replace(' ','_')
dataset = dataset.rename(columns = {'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
dataset.isnull().sum()
```

```python
#creating a new age group:
labels = ['young adult' , 'Adult' , 'Middle Age' , 'Senior']

dataset['age_group'] = pd.qcut(dataset['age'] , q=4 , labels = labels)
```

```python
dataset[['age' , 'age_group']].head(20)
```

```python
#frequency mapping

frequency_mapping = {
    
'Fortnightly': 14,
'Weekly' : 7,
'Monthly' : 30,
'Quarterly' : 90,
'Bi-Weekly' : 14,
'Annually' : 365,
'Every 3 Months' : 90
}

dataset['purchase_frequency_days'] = dataset['frequency_of_purchases'].map(frequency_mapping)
```

```python
dataset[['purchase_frequency_days' , 'frequency_of_purchases']].head(20)
```

```python
#checking if promo_codes and discount_used are similar

dataset[['promo_code_used' , 'discount_applied']].head(20)
```

```python
(dataset['promo_code_used'] == dataset['discount_applied']).all()
```

```python
#no use in having the same table again
dataset = dataset.drop('promo_code_used' , axis = 1)
```

```python
import mysql.connector
import pandas as pd

from sqlalchemy import create_engine, text

# MySQL credentials
username = "root"
password = "root"
host = "127.0.0.1"
port = "3306"
database = "customer_behavior"

print("Connected successfully!")

# Create engine
engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
)

# Upload dataframe to MySQL
table_name = "customer"

dataset.to_sql(
    table_name,
    engine,
    if_exists="replace",
    index=False
)

# Read data back
with engine.connect() as conn:

    query = text("SELECT * FROM customer LIMIT 5;")

    result = pd.read_sql(
        query,
        conn
    )

print(result)
```

```python
with engine.connect() as conn:

    query = text("SELECT location from customer ;")

    result = pd.read_sql(
        query,
        conn
    )

print(result)
```
